```python
import pandas as pd
energie = pd.read_csv("eco2mix.csv", sep = ";")
energie.head()
```

```python
#Analyse de la colonne TCO Thermique
print("le nombre de ligne NA est de ",energie["TCO Nucléaire (%)"].isna().sum())
NbrNA = energie["TCO Nucléaire (%)"].isna().sum()
print("% de valeur NA dans la colonne :", (NbrNA/energie.shape[0])*100)

#Vérification des valeurs maximales et minimales
print("valeur minimale : ",energie["TCO Nucléaire (%)"].min())
print("valeur maximale : ",energie["TCO Nucléaire (%)"].max())
```

```python
#Vérification des colonne TCO Nucléaire avec une valeur de 0
filtre = energie.loc[energie["TCO Nucléaire (%)"]==0]
print("Il y a",filtre.shape[0],"ligne avec un TCO Nucléaire égale à 0")

print("Il y a ",(filtre["Nucléaire (MW)"]==0).sum(),"ligne avec une valeur 0 pour Nucléaire MW quand Nucléaire TCO est égale à 0")
```
