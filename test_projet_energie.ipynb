```python
import pandas as pd
```

```python
energie = pd.read_csv("eco2mix-regional-cons-def.csv", sep = ";")
display(energie.head())

energie_filtre = energie[["Région", "Date", "Heure", "Consommation (MW)", "Solaire (MW)", "TCO Solaire (%)", "TCH Solaire (%)" ]]

print(energie_filtre.isna().sum())
print(energie_filtre.dtypes)
```

```python
#Recherche des causes des NA dans la colonne "TCO Solaire (%)" avec les colonnes "Consommation (MW)" et "Solaire (MW)"
# (puisque TCO Solaire = Solaire / Consommation)

total_tco_na = energie_filtre['TCO Solaire (%)'].isna().sum()
print("il y a au total ", total_tco_na, " valeurs manquantes pour la colonne 'TCO Solaire (%)'", "\n")


# Pour toutes les lignes NA de TCO Solaire, on regarde ce qui se passe dans les autres colonnes

#On filtre sur les NA de TCO Solaire, puis sur la colonne Solaire. Filtre stocké dans subset_solaire
subset_solaire = energie_filtre[(energie_filtre['TCO Solaire (%)'].isna())]['Solaire (MW)']

#Colonne production solaire : on vérifie s'il y a des NA
nb_na_solaire = subset_solaire.isna().sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_na_solaire, " NA dans la colonne Solaire" )

#Colonne production solaire : on vérifie s'il y a des zéros
nb_zero_solaire = (subset_solaire == 0).sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_zero_solaire, " zéros dans la colonne Solaire" )

#Colonne production solaire : On vérifie s'il y a des valeurs négatives
nb_neg_solaire = (subset_solaire<0).sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_neg_solaire, " valeurs négatives dans la colonne Solaire", "\n" )


#On filtre sur les NA de TCO Solaire, puis sur la colonne Consommation. Filtre stocké dans subset_conso
subset_conso = energie_filtre[(energie_filtre['TCO Solaire (%)'].isna())]['Consommation (MW)']

#Colonne production conso : on vérifie s'il y a des NA
nb_na_conso = subset_conso.isna().sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_na_conso, " NA dans la colonne 'Consommation'" )

#Colonne production conso : on vérifie s'il y a des zéros
nb_zero_conso = (subset_conso == 0).sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_zero_conso, " zéros dans la colonne 'Consommation'" )

#Colonne production conso : On vérifie s'il y a des valeurs négatives
nb_neg_conso = (subset_conso<0).sum()
print("quand on filtre sur les NA de TCO Solaire, il y a ", nb_neg_conso, " valeurs négatives dans la colonne 'Consommation'", "\n" )

print("test solaire : ", total_tco_na == nb_na_solaire+nb_zero_solaire+nb_neg_solaire)
print("test conso : ", total_tco_na == nb_na_conso+nb_zero_conso+nb_neg_conso)
```

```python
#Recherche des causes des NA dans la colonne "TCO Solaire (%)" avec la colonne "Date"

energie_filtre['Date'] = pd.to_datetime(energie_filtre['Date'])
energie_filtre = energie_filtre.sort_values(by='Date', ascending=True)
total_tco_na = 1472257

# 1. Regarder les 1472257 premières lignes du df trié (positions 0 à 1472256 inclus, donc les dates les plus anciennes),
# pour voir si pour ces lignes la colonne 'TCO Solaire' contient des NAN
subset_date = energie_filtre.head(total_tco_na)['TCO Solaire (%)'].isna().sum()
print("1. nombre de NAN dans la colonne 'TCO Solaire' lorsque l'on filtre sur les 1472257 lignes les plus anciennes du df : ", subset_date)
print("1.bis. nombre de non-NAN dans la colonne 'TCO Solaire', avec ce même filtre : ", total_tco_na-subset_date, "\n")



# 3. On vérifie que la ligne 1472257 (juste après le bloc précédent) a bien une valeur dans TCO Solaire, et pas des non-NaN
subset_1472257 = pd.isna(energie_filtre.iloc[1472257]['TCO Solaire (%)'])
print("3. la ligne 1472257 contient-elle un NAN (si False, c'est OK) :", subset_1472257, "\n")

# 4. On vérifie que toutes les lignes suivantes (à partir de la ligne 1472258) ont bien une valeur TCO Solaire, et pas des non-NaN
subset_dates_suivantes = pd.isna(energie_filtre.iloc[1472258:]['TCO Solaire (%)']).sum()
print("4. nombre de NAN dans la colonne 'TCO Solaire' lorsque l'on filtre à partir de la 1472258eme ligne du df (dates plus récentes) : ", subset_dates_suivantes, "n")


```

```python
# Pour toutes les lignes NA de TCO Solaire, on vérifie si toutes les valeurs de TCH Solaire sont des NA aussi.

total_tch_na = energie_filtre['TCH Solaire (%)'].isna().sum()
print("nombre de NA dans la colonne tch solaire : ",total_tch_na)

# On filtre sur les NA pour TCO solaire, et on regarde le total de NA dans la TCH Solaire correspondante
test = energie_filtre[(energie_filtre['TCO Solaire (%)'].isna())]['TCH Solaire (%)'].isna().sum()
print("nombre de NA dans la colonne TCH solaire lorsque la colonne TCO contient aussi des NA : ", test)

print(total_tch_na == test)

```
