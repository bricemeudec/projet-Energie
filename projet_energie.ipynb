```python
#Analyse de la colonne TCO Thermique
print("le nombre de ligne NA est de ",energie["TCO Thermique (%)"].isna().sum())
NbrNA = energie["TCO Thermique (%)"].isna().sum()
print("% de valeur NA dans la colonne :", (NbrNA/energie.shape[0])*100)

#Vérification des valeurs maximales et minimales
print("valeur minimale : ",energie["TCO Thermique (%)"].min())
print("valeur maximale : ",energie["TCO Thermique (%)"].max())
```

```python
#Vérification des colonne TCO Thermique avec une valeur de 0
filtre = energie.loc[energie["TCO Thermique (%)"]==0]
print("Il y a",filtre.shape[0],"ligne avec un TCO Thermique égale à 0")

print("Il y a ",(filtre["Thermique (MW)"]==0).sum(),"ligne avec une valeur 0 pour Thermique MW quand Thermique TCO est égale à 0")
```

```python
#Analyse de la colonne TCH Thermique
print("le nombre de ligne NA est de ",energie["TCH Thermique (%)"].isna().sum())
NbrNA = energie["TCH Thermique (%)"].isna().sum()
print("% de valeur NA dans la colonne :", (NbrNA/energie.shape[0])*100)

#Vérification des valeurs maximales et minimales
print("valeur minimale : ",energie["TCH Thermique (%)"].min())
print("valeur maximale : ",energie["TCH Thermique (%)"].max())

#Vérification des colonne TCO Thermique avec une valeur de 0
filtre = energie.loc[energie["TCH Thermique (%)"]==0]
print("Il y a",filtre.shape[0],"ligne avec un TCH Thermique égale à 0")

print("Il y a ",(filtre["Thermique (MW)"]==0).sum(),"ligne avec une valeur 0 pour Thermique MW quand Thermique TCH est égale à 0")
```
