```python

energie_filtre_hydraulique = energie[["Région", "Date", "Heure", "Hydraulique (MW)", "TCO Hydraulique (%)", "TCH Hydraulique (%)" ]]

print(energie_filtre_hydraulique.isna().sum())
print(energie_filtre_hydraulique.dtypes)
```

```python
# Synthèse des valeurs manquantes par colonne
na_summary = pd.DataFrame({
    'nombre_na': energie.isna().sum(),
    'taux_na_%': (energie.isna().mean() * 100).round(2)
}).sort_values(by='taux_na_%', ascending=False)

print(na_summary)
```

```python
#print("Valeurs possibles de la colonnes Eolien (MW) :",energie['Eolien (MW)'].unique())
# Conversion simple en entier
# 1. Remplacer la virgule par un point et supprimer les espaces si nécessaire
energie['Eolien (MW)'] = energie['Eolien (MW)'].astype(str).str.replace(',', '.').str.strip()

# 2. Convertir en numérique (force les erreurs/textes invalides en NaN)
energie['Eolien (MW)'] = pd.to_numeric(energie['Eolien (MW)'], errors='coerce')

# 3. Transformer en entier nullable
energie['Eolien (MW)'] = energie['Eolien (MW)'].astype('Int64')
```

```python
sns.relplot(x=energie['Région'], y=energie['Hydraulique (MW)'])
fig = plt.gcf()
fig.set_size_inches(15, 10)
plt.title('Répartition des données Hydraulique (MW) par région')
plt.xlabel('Région')
plt.ylabel('Hydraulique (MW)')
plt.xticks(rotation=90)
plt.show()
```
